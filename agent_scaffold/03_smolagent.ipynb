# test-3
除了按照 Smolagents 规定的格式编写本地工具外，Smolagents 还封装了 MCP Client，可以连接 Stdio 与 Streamable-HTTP 协议的 MCP Server，从而调用工具。

用CC编写 mcp_tools.py 中2个tool函数，提示词：
```
在 @mcp_tools.py 文件中写两个函数：
第一个是read_csv函数，用于读取指定CSV文件的内容，第二个函数是write_md函数，用于将markdown内容写入到指定的md文件中。
之后将这两个函数，使用mcp库的FastMCP包，封装为一个MCP Server的两个工具，MCP Server使用stream-http方式通信，host为0.0.0.0，端口为38000
```

先启动 mcp server.(我在 vs code中手动启动)

> 代码修改处：开启stream.
若项目使用的模型供应商为modelscope, 对于 reasoning模型，e.g. glm-5.2, 在非流式下，api经常返回 choices=null,但stream模式下正常。

我先用modelscope,但跑得太慢，后切换为硅基流动

````python
from smolagents import CodeAgent, MCPClient, OpenAIModel
from dotenv import load_dotenv
import os

load_dotenv(override=True)
model = OpenAIModel(
    model_id="zai-org/GLM-5.2"
)

with MCPClient({"url": "http://127.0.0.1:38000/mcp", "transport": "streamable-http"}) as tools:
    agent = CodeAgent(tools=tools, model=model, stream_outputs=True,
                      additional_authorized_imports=["*"])#此处允许沙箱执行任意操作是危险的
    
    prompt = """
    针对`03-04 Smolagents/yanjing_beer_daily_k_20250518_20260518.csv`(文件位于mcp server)中的数据展开走势分析，
    并输出一份markdown格式的分析报告，写入到 `data/report_smolagents_test3_qwen3.md` (本地目录)文件中.
    最终报告需要根据真实数据，不要编造报告。
    若代码执行过程中缺少pip包，可以自行安装，本项目使用了 uv 创建虚拟环境。
    """

    agent.run(prompt)
````
