本节详细代码见目录 08_subagent.

| 概念 | overview |
|---|---|
| 为什么要多智能体 | 拆分职责，避免单 Agent 上下文污染与权限过大 |
| Orchestrator-Worker | 主 Agent 分发，子 Agent 专职执行，结果汇总回主 Agent |
| 并行扇出 | 独立子任务同时跑，耗时=最慢的一个 |
| 上下文隔离 | 子 Agent 全新上下文，只把最终结果带回父层 |
| 工具限制 | 按最小权限原则给每个子 Agent 分配工具 |
| description 字段 | 主 Agent 靠它做"任务-Agent"的自动匹配 |
| Agent 工具 | 主 Agent 调用子 Agent 的机制，需加入 allowed_tools |
| 何时不用 subagent 而用 Workflow | 需要协调几十到上百个 Agent 时，用 Workflow 把编排搬到脚本层，避免撑爆对话上下文 |
